# BCG X — GFC AI Chatbot Project
## Task 1: Financial Data Extraction & Analysis
**Companies:** Microsoft, Tesla, Apple | **Period:** Last 3 Fiscal Years

```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Load data
df = pd.read_csv('financial_data.csv')
print('Shape:', df.shape)
df.head(10)
```

## 1. Data Overview & Cleaning

```python
print('=== Data Types ===')
print(df.dtypes)
print('\n=== Missing Values ===')
print(df.isnull().sum())
print('\n=== Basic Statistics ===')
df.describe().round(2)
```

## 2. Year-over-Year Growth Calculations

```python
df = df.sort_values(['Company', 'Year']).reset_index(drop=True)

# YoY % changes per company
df['Revenue_Growth (%)']    = df.groupby('Company')['Total_Revenue'].pct_change() * 100
df['NetIncome_Growth (%)']  = df.groupby('Company')['Net_Income'].pct_change() * 100
df['Assets_Growth (%)']     = df.groupby('Company')['Total_Assets'].pct_change() * 100
df['OpCF_Growth (%)']       = df.groupby('Company')['Operating_CF'].pct_change() * 100

# Derived ratios
df['Net_Profit_Margin (%)'] = (df['Net_Income'] / df['Total_Revenue']) * 100
df['Debt_Ratio']            = df['Total_Liabilities'] / df['Total_Assets']
df['OpCF_to_Revenue (%)']   = (df['Operating_CF'] / df['Total_Revenue']) * 100

print('=== Enriched Dataset ===')
df.round(2)
```

## 3. Company-Level Summary

```python
summary = df.groupby('Company').agg(
    Avg_Revenue      = ('Total_Revenue',       'mean'),
    Avg_Net_Income   = ('Net_Income',           'mean'),
    Avg_Revenue_Grwth= ('Revenue_Growth (%)',   'mean'),
    Avg_NI_Growth    = ('NetIncome_Growth (%)','mean'),
    Avg_Margin       = ('Net_Profit_Margin (%)','mean'),
    Avg_Debt_Ratio   = ('Debt_Ratio',           'mean'),
    Avg_OpCF         = ('Operating_CF',         'mean')
).round(2)
print('=== Company Summary ===')
summary
```

## 4. Visualisations

```python
companies = df['Company'].unique()
colors    = {'Microsoft': '#0078D4', 'Tesla': '#CC0000', 'Apple': '#555555'}

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('GFC AI Chatbot Project — Financial Analysis Dashboard', fontsize=15, fontweight='bold')

metrics = [
    ('Total_Revenue',       'Total Revenue (B USD)',        axes[0,0]),
    ('Net_Income',          'Net Income (B USD)',           axes[0,1]),
    ('Operating_CF',        'Operating Cash Flow (B USD)',  axes[0,2]),
    ('Net_Profit_Margin (%)', 'Net Profit Margin (%)',      axes[1,0]),
    ('Debt_Ratio',          'Debt Ratio',                  axes[1,1]),
    ('Revenue_Growth (%)',  'Revenue Growth YoY (%)',       axes[1,2]),
]

for col, title, ax in metrics:
    for company in companies:
        sub = df[df['Company'] == company]
        ax.plot(sub['Year'], sub[col], marker='o', label=company, color=colors[company], linewidth=2)
    ax.set_title(title, fontsize=11)
    ax.set_xlabel('Year')
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('financial_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()
print('Chart saved.')
```

## 5. Key Findings & Insights

```python
print('=' * 60)
print('KEY FINDINGS FOR GFC AI CHATBOT PROJECT')
print('=' * 60)

for company in companies:
    sub = df[df['Company'] == company].dropna(subset=['Revenue_Growth (%)'])
    avg_rev_growth = sub['Revenue_Growth (%)'].mean()
    avg_margin     = df[df['Company'] == company]['Net_Profit_Margin (%)'].mean()
    avg_debt       = df[df['Company'] == company]['Debt_Ratio'].mean()
    latest_rev     = df[df['Company'] == company]['Total_Revenue'].iloc[-1]

    print(f'\n--- {company} ---')
    print(f'  Latest Revenue      : ${latest_rev:.1f}B')
    print(f'  Avg Revenue Growth  : {avg_rev_growth:.1f}%')
    print(f'  Avg Net Margin      : {avg_margin:.1f}%')
    print(f'  Avg Debt Ratio      : {avg_debt:.2f}')

print('\n' + '=' * 60)
print('SUMMARY FOR AI MODEL INTEGRATION')
print('=' * 60)
print('Data is clean, structured, and ready for the GFC chatbot.')
print('Derived features: growth rates, profit margin, debt ratio, OpCF yield.')
print('Recommended next step: feed this enriched DataFrame into the AI model.')
```

## 6. Export Enriched Data

```python
df.round(2).to_csv('financial_data_enriched.csv', index=False)
print('Enriched CSV saved: financial_data_enriched.csv')
```
